```python
# ==============================================================
# Employee Management System (EMS)
# ==============================================================


# --------------------------------------------------------------
# Step 1: Plan the Data Storage
# Employees are stored in a dictionary. The key is the emp_id and
# the value is another dictionary with name, age, department, salary.
# --------------------------------------------------------------
employees = {
    101: {'name': 'Satya', 'age': 27, 'department': 'HR', 'salary': 50000},
    102: {'name': 'Rahul', 'age': 30, 'department': 'IT', 'salary': 70000},
}


# --------------------------------------------------------------
# Step 2: Define the Menu System
# main_menu() shows the options in a loop and calls the matching
# function. The loop stops only when the user chooses Exit.
# --------------------------------------------------------------
def main_menu():
    while True:
        print("\n===== Employee Management System =====")
        print("1. Add Employee")
        print("2. View All Employees")
        print("3. Search for Employee")
        print("4. Exit")

        choice = input("Enter your choice (1-4): ")

        if choice == "1":
            add_employee()
        elif choice == "2":
            view_employees()
        elif choice == "3":
            search_employee()
        elif choice == "4":
            print("Thank you for using the Employee Management System. Goodbye!")
            break
        else:
            print("Invalid choice. Please enter a number from 1 to 4.")


# --------------------------------------------------------------
# Step 3: Add Employee Functionality
# Takes the employee details from the user. The Employee ID must
# be unique, otherwise the user is asked to enter a new one.
# The employee is then saved in the dictionary.
# --------------------------------------------------------------
def add_employee():
    print("\n--- Add Employee ---")

    while True:
        emp_id = input("Enter Employee ID: ")
        if not emp_id.isdigit():
            print("ID must be a number. Try again.")
        elif int(emp_id) in employees:
            print("This ID already exists. Please enter a new ID.")
        else:
            emp_id = int(emp_id)
            break

    name = input("Enter Employee Name: ")

    age = input("Enter Employee Age: ")
    while not age.isdigit():
        print("Age must be a number. Try again.")
        age = input("Enter Employee Age: ")
    age = int(age)

    department = input("Enter Employee Department: ")

    salary = input("Enter Employee Salary: ")
    while not salary.isdigit():
        print("Salary must be a number. Try again.")
        salary = input("Enter Employee Salary: ")
    salary = int(salary)

    employees[emp_id] = {
        'name': name,
        'age': age,
        'department': department,
        'salary': salary
    }

    print("Employee added successfully!")


# --------------------------------------------------------------
# Step 4: View All Employees
# Shows every employee in a table. If the dictionary is empty,
# a message is displayed instead.
# --------------------------------------------------------------
def view_employees():
    print("\n--- All Employees ---")

    if len(employees) == 0:
        print("No employees available.")
        return

    print(f"{'ID':<8}{'Name':<15}{'Age':<8}{'Department':<15}{'Salary':<10}")
    print("-" * 56)

    for emp_id in employees:
        emp = employees[emp_id]
        print(f"{emp_id:<8}{emp['name']:<15}{emp['age']:<8}{emp['department']:<15}{emp['salary']:<10}")


# --------------------------------------------------------------
# Step 5: Search for an Employee by ID
# Asks for an Employee ID and shows the details if it exists,
# otherwise displays "Employee not found."
# --------------------------------------------------------------
def search_employee():
    print("\n--- Search Employee ---")
    emp_id = input("Enter Employee ID to search: ")

    if emp_id.isdigit() and int(emp_id) in employees:
        emp = employees[int(emp_id)]
        print("Employee found!")
        print("Name       :", emp['name'])
        print("Age        :", emp['age'])
        print("Department :", emp['department'])
        print("Salary     :", emp['salary'])
    else:
        print("Employee not found.")


# --------------------------------------------------------------
# Step 6: Exit the Program
# Choosing option 4 in main_menu() prints a thank-you message
# and ends the loop. The line below starts the program.
# --------------------------------------------------------------
main_menu()
```
